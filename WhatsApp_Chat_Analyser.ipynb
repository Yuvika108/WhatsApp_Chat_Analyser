```python
import re
import pandas as pd
```

```python
with open("WhatsApp Chat with Team Networking and Outreach (2025-26).txt", "r", encoding="utf-8") as f:
    data = f.read()
```

```python
# Pattern matched to your exact file format
pattern = r'(\d{1,2}/\d{2}/\d{2}, \d{2}:\d{2}) - (.*?): (.*)'
matches = re.findall(pattern, data)
```

```python
# Build DataFrame
df = pd.DataFrame({
    'message_date': [m[0] for m in matches],
    'user':         [m[1] for m in matches],
    'message':      [m[2] for m in matches]
})
```

```python
dates        = [m[0] for m in matches]
users_raw    = [m[1] for m in matches]
messages_raw = [m[2] for m in matches]
```

```python
# Build DataFrame
df = pd.DataFrame({
    'message_date':    dates,
    'user_message_raw': users_raw,
    'user_message':    messages_raw
})
```

```python
# Convert date
df['date'] = pd.to_datetime(df['message_date'], format='%d/%m/%y, %H:%M')
df.drop(columns=['message_date'], inplace=True)
```

```python
# Time columns
df['year']  = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day']   = df['date'].dt.day
df['hour']  = df['date'].dt.hour

print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```
